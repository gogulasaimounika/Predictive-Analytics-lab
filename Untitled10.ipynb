```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import LabelEncoder
import numpy as np

# Load your dataset
df = pd.read_csv("retail_sales_dataset.csv")

# Use Customer ID and Product Category only
df = df[['Customer ID', 'Product Category']]

# Create a binary matrix: rows=customers, columns=product categories
user_item_matrix = df.pivot_table(index='Customer ID',
                                  columns='Product Category',
                                  aggfunc=lambda x: 1,
                                  fill_value=0)

# Compute cosine similarity between customers
customer_similarity = cosine_similarity(user_item_matrix)
customer_similarity_df = pd.DataFrame(customer_similarity,
                                      index=user_item_matrix.index,
                                      columns=user_item_matrix.index)

# Recommend products for a given customer
def recommend_products(customer_id, top_n=3):
    if customer_id not in customer_similarity_df:
        return f"Customer ID {customer_id} not found."

    # Get similar customers
    similar_scores = customer_similarity_df[customer_id].sort_values(ascending=False)[1:]  # exclude self
    similar_customers = similar_scores.index

    # Products this customer already bought
    customer_products = set(df[df['Customer ID'] == customer_id]['Product Category'])

    # Get potential recommendations
    recommendations = {}
    for other_customer in similar_customers:
        other_products = set(df[df['Customer ID'] == other_customer]['Product Category'])
        for product in other_products - customer_products:
            recommendations[product] = recommendations.get(product, 0) + 1

        if len(recommendations) >= top_n:
            break

    # Sort and return top N recommendations
    sorted_recs = sorted(recommendations.items(), key=lambda x: x[1], reverse=True)
    return [prod for prod, _ in sorted_recs[:top_n]]

# Example usage
print("Recommended for CUST001:", recommend_products('CUST010'))
```
